# 二次函数实验场

在这里，公式、代码和图像可以一起运行。我们将研究：

$$f(x)=ax^2+bx+c$$

先选择菜单 **运行 → 全部运行**，然后滚动到最后的交互实验区，拖动滑块观察图像、判别式和零点的变化。

你也可以修改任何代码，再按 **Shift + Enter** 重新运行当前单元格。

```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
from IPython.display import display, Math
from ipywidgets import FloatSlider, interact

print('实验环境准备好了。')
```

## 1. 让 Python 求根

先用 SymPy 解方程 $x^2-2x-3=0$。把下面的表达式换成你的题目，再运行一次即可。

```python
x = sp.symbols('x', real=True)
expression = x**2 - 2*x - 3
roots = sp.solve(expression, x)
display(Math('方程：' + sp.latex(sp.Eq(expression, 0))))
display(Math('根：' + sp.latex(roots)))
```

## 2. 拖动滑块做实验

- 改变 $a$：观察开口方向和开口大小。
- 改变 $b$：观察对称轴的位置。
- 改变 $c$：观察图像上下移动。
- 让判别式 $\Delta=b^2-4ac$ 分别大于、等于、小于零，观察零点数量。

```python
def quadratic_lab(a=1.0, b=-2.0, c=-3.0):
    x_values = np.linspace(-8, 8, 600)
    y_values = a*x_values**2 + b*x_values + c

    fig, ax = plt.subplots(figsize=(9, 5))
    ax.plot(x_values, y_values, color='#6575d8', linewidth=2.5, label='f(x)')
    ax.axhline(0, color='#777780', linewidth=0.8)
    ax.axvline(0, color='#777780', linewidth=0.8)
    ax.grid(alpha=0.18)
    ax.set_xlabel('x')
    ax.set_ylabel('f(x)')

    display(Math(rf'f(x)={a:g}x^2 {b:+g}x {c:+g}'))

    if abs(a) < 1e-12:
        display(Math('这是一次函数（或常数函数），因为 a=0。'))
        if abs(b) > 1e-12:
            root = -c/b
            ax.scatter([root], [0], color='#d45d79', s=60, zorder=3)
            display(Math(rf'唯一零点：x={root:g}'))
    else:
        delta = b**2 - 4*a*c
        vertex_x = -b/(2*a)
        vertex_y = a*vertex_x**2 + b*vertex_x + c
        ax.scatter([vertex_x], [vertex_y], color='#43a58a', s=50, zorder=3, label='顶点')
        ax.axvline(vertex_x, color='#43a58a', linewidth=1, linestyle='--', alpha=0.6)
        display(Math(rf'\Delta=b^2-4ac={delta:g},\quad 顶点=({vertex_x:g},{vertex_y:g})'))

        if delta > 0:
            roots = ((-b-np.sqrt(delta))/(2*a), (-b+np.sqrt(delta))/(2*a))
            ax.scatter(roots, [0, 0], color='#d45d79', s=60, zorder=3, label='零点')
            display(Math(rf'两个实根：x_1={roots[0]:g},\quad x_2={roots[1]:g}'))
        elif abs(delta) < 1e-12:
            root = -b/(2*a)
            ax.scatter([root], [0], color='#d45d79', s=60, zorder=3, label='零点')
            display(Math(rf'两个相同实根：x={root:g}'))
        else:
            display(Math('没有实根：抛物线与 x 轴不相交。'))

    ax.legend()
    plt.show()

interact(
    quadratic_lab,
    a=FloatSlider(value=1.0, min=-3.0, max=3.0, step=0.1, description='a'),
    b=FloatSlider(value=-2.0, min=-6.0, max=6.0, step=0.1, description='b'),
    c=FloatSlider(value=-3.0, min=-8.0, max=8.0, step=0.1, description='c'),
);
```
